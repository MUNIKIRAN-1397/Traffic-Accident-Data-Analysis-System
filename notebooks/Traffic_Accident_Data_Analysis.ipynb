```python
# ============================================================
# TRAFFIC ACCIDENT DATA ANALYSIS SYSTEM
# Data Analysis Essentials - Cornerstone Project
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES AND LOAD DATASET
# ============================================================

import numpy as np
import pandas as pd

from google.colab import files
uploaded = files.upload()

df = pd.read_csv("traffic_accidents_raw.csv")

print("Dataset shape before cleaning:",
      df.shape[0], "rows x", df.shape[1], "columns")
print("\n")


# ============================================================
# 2. INITIAL DATA INSPECTION
# ============================================================

# First five records
print("========== FIRST 5 RECORDS ==========")
print(df.head())
print("\n")


# Data types
print("========== DATA TYPES ==========")
print(df.dtypes)
print("\n")


# ============================================================
# 3. CATEGORICAL DATA INSPECTION
# ============================================================

categorical_columns = [
    "Accident_Severity",
    "Vehicle_Type",
    "Weather_Condition",
    "Road_Type",
    "Light_Condition"
]

for col in categorical_columns:
    print("==========", col, "CATEGORIES ==========")
    print(df[col].unique())
    print()

# ============================================================
# 4. CHECK DUPLICATE AND MISSING VALUES
# ============================================================

# Duplicate records
print("========== DUPLICATE RECORDS ==========")
print("Duplicate rows detected:", df.duplicated().sum())
print("\n")

# Missing values
print("========== MISSING VALUES ==========")
print("Missing values before cleaning:",
      df.isnull().sum().sum())
print("\n")


# ============================================================
# 5. DATA CLEANING
# ============================================================

# Remove duplicate records
df = df.drop_duplicates()

# Fill missing categorical values using Mode
for col in ["Light_Condition", "Weather_Condition"]:
    df[col] = df[col].fillna(df[col].mode()[0])

# Fill missing numerical values using Median
for col in ["Number_of_Casualties", "Number_of_Fatalities"]:
    df[col] = df[col].fillna(df[col].median())


# ============================================================
# 6. DATA CLEANING RESULTS
# ============================================================

print("========== AFTER CLEANING ==========")

print("Rows              :", df.shape[0])
print("Columns           :", df.shape[1])
print("Missing values    :", df.isnull().sum().sum())
print("Duplicate rows    :", df.duplicated().sum())
print("\n")

print("First 5 records after cleaning:")
print(df.head())
print("\n")


# ============================================================
# 7. DATA PREPROCESSING
# ============================================================

from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder


# ============================================================
# 7.1 FEATURE SCALING
# ============================================================

scaler = MinMaxScaler()

df["Speed_Limit (km/h)"] = scaler.fit_transform(
    df[["Speed_Limit (km/h)"]]
)

print("========== AFTER FEATURE SCALING ==========")
print(df[["Speed_Limit (km/h)"]].head())
print("\n")


# ============================================================
# 7.2 FEATURE STANDARDIZATION
# ============================================================

standard = StandardScaler()

df["Number_of_Casualties"] = standard.fit_transform(
    df[["Number_of_Casualties"]]
)

df["Number_of_Fatalities"] = standard.fit_transform(
    df[["Number_of_Fatalities"]]
)

print("========== AFTER FEATURE STANDARDIZATION ==========")
print(df[[
    "Number_of_Casualties",
    "Number_of_Fatalities"
]].head())
print("\n")


# ============================================================
# 7.3 LABEL ENCODING
# ============================================================

encoder = LabelEncoder()

df["Accident_Severity"] = encoder.fit_transform(
    df["Accident_Severity"]
)

print("========== AFTER LABEL ENCODING ==========")
print(df[["Accident_Severity"]].head())
print("\n")


# Label Mapping
print("========== ACCIDENT SEVERITY LABEL MAPPING ==========")

for i, label in enumerate(encoder.classes_):
    print(label, "=", i)

print("\n")


# ============================================================
# 7.4 ONE-HOT ENCODING
# ============================================================

df = pd.get_dummies(
    df,
    columns=["Vehicle_Type"]
)

print("========== AFTER ONE-HOT ENCODING ==========")
print(df.head())
print("\n")


# ============================================================
# 8. FINAL PREPROCESSED DATASET
# ============================================================

print("========== FINAL PREPROCESSED DATASET ==========")
print(df.head())
```
